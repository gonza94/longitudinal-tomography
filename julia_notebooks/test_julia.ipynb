```julia
using LinearAlgebra

####################################################
# Hello Julia!
####################################################

println("Hello, Julia!")

# Variables
x = 3.0
y = 4

println("x = $x")
println("y = $y")
println("x + y = $(x + y)")
```

```julia
####################################################
# Arrays
####################################################

a = [1,2,3,4,5]

println("\nArray:")
println(a)

# Julia arrays start at 1
println("First element = ", a[1])
```

```julia
####################################################
# Matrix
####################################################

A = rand(4,4)

println("\nRandom matrix:")
println(A)

println("\nEigenvalues:")
println(eigvals(A))
```

```julia
####################################################
# Functions
####################################################

function f(x)
    return x^2 + 2x + 1
end

println("\nf(3) = ", f(3))

# One-line function
g(x) = sin(x)

println("g(pi/2) = ", g(pi/2))

```

```julia
####################################################
# Broadcasting
####################################################

xs = range(-2,2,length=10)

println("\nBroadcasting")

println(f.(xs))
println(g.(xs))

```

```julia
####################################################
# Loops
####################################################

println("\nSquares")

for i in 1:5
    println("$i -> $(i^2)")
end

```

```julia
####################################################
# List comprehensions
####################################################

squares = [i^2 for i in 1:10]

println("\nSquares array:")
println(squares)
```

```julia
####################################################
# Conditionals
####################################################

value = -2

if value > 0
    println("positive")
elseif value < 0
    println("negative")
else
    println("zero")
end
```

```julia
using Statistics
####################################################
# Multiple return values
####################################################

function statistics(v)
    return mean(v), std(v)
end

μ, σ = statistics(randn(1000))

println("\nMean = $μ")
println("Std  = $σ")
```

```julia
using Plots

x = range(-2π,2π,length=500)

plot(
    x,
    [sin.(x), cos.(x)],
    xlabel="x",
    ylabel="f(x)",
    label=["sin" "cos"],
    linewidth=2,
    title="Julia plotting"
)
```

```julia
using LinearAlgebra

N = 1000

A = randn(N,N)

println("Computing SVD...")

@time U,S,V = svd(A)

println("Largest singular value = ", maximum(S))
```

```julia
# using Pkg
# Pkg.add("ForwardDiff")

using ForwardDiff

f(x) = sin(x)^2 + x^3

println(f(2.0))

println(ForwardDiff.derivative(f,2.0))
```

```julia
using Plots

dt = 0.01
T = 20.0

x = 1.0
v = 0.0

xs = Float64[]
ts = Float64[]

for t in 0:dt:T

    push!(xs,x)
    push!(ts,t)

    a = -x

    x += dt*v
    v += dt*a

end

plot(ts,xs,
    xlabel="Time",
    ylabel="Position",
    label="x(t)",
    linewidth=2)
```
